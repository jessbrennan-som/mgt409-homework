Verifies the environment: imports the shared `mgt409` helpers plus pandas/matplotlib and renders a chart inline.

```python
import pandas as pd
import matplotlib.pyplot as plt
from mgt409 import describe, moving_average, summary_table
print('mgt409 helpers imported OK')
```

```python
sales = [95.5, 99.6, 109.8, 125.5, 128.2, 120.9, 122.7, 144.4, 134.6, 139.2, 146.9, 155.6]
months = pd.date_range('2025-01-01', periods=12, freq='MS')
df = pd.DataFrame({'month': months, 'sales': sales})
describe(df['sales'])
```

```python
window = 3
ma = moving_average(df['sales'], window)
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(df['month'], df['sales'], marker='o', label='Monthly sales')
ax.plot(df['month'].iloc[window-1:], ma, color='crimson', lw=2, label=f'{window}-mo moving avg')
ax.set_title('MGT 409 — Monthly Sales')
ax.legend(); ax.grid(alpha=0.3)
plt.show()
```
